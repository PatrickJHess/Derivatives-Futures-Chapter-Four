In previous chapters and notebooks, we examined Stock Index, SOFR, and Fed Funds futures. These contracts share a common trait: Cash Settlement. At expiration, the final settlement is calculated from the difference between the final futures price and the reference index.

Cash settlement requires a transparent reference value—a value that doesn't exist for Treasury futures. The cash Treasury market is Over-The-Counter (OTC) and surprisingly murky. The newest "On-The-Run" Treasury bonds are liquid, but thousands of older "Off-The-Run" bonds are not. Bonds that don't trade regularly are unreliable reference points.

Physical delivery solves the problem. A short must deliver a bond to close out a position after trading stops (the music stops 7 business days before the last business day of the contract month). By forcing the short seller to actually acquire an eligible bond in the secondary market, the futures price is linked to the cash market. Importantly, the settlement mechanism includes a menu of eligible bonds that ensures a strong link between the cash and futures market.


## 📜 Mapping the Treasury Curve: The Futures Menu
To allow market participants to hedge interest rate risk across the entire yield curve, the CME lists specific contracts targeting the short-end, the belly, and the long-end of the curve. The five primary physical delivery contracts and their root symbols (excluding the expiration month and year) are:
1. **2-Year Treasury Notes (ZT):** Covers the short end. Original maturity not greater than 5 years and 3 months, with a remaining maturity of at least 1 year and 9 months.
2. **5-Year Treasury Notes (ZF):** Covers the belly of the curve. Original maturity not greater than 5 years and three months, with a remaining maturity of at least 4 years and 2 months.
3. **10-Year Treasury Notes (ZN):** Covers the benchmark 10-year maturity. Eligible bonds must have a remaining maturity between 6.5 and 8 years.
4. **Treasury Bonds (The "Classic" Bond) (ZBZ:** Covers the long end. Eligible bonds must have a remaining maturity between 15 and 25 years.
5. **Ultra Treasury Bonds (UB):** Covers the extreme long end, accepting bonds with a remaining maturity of at least 25 years.
)

## ⚖️ The Conversion Factor: Leveling the Playing Field
If the exchange allowed the short seller to deliver any eligible bond in the basket without adjustment for 'quality', the market would instantly break. Short sellers would deliver the bond with the lowest absolute cash price—invariably the bond with the lowest coupon. The result would be a delivery menu of one, two, or three bonds. The shorter the delivery menu, the more limited the hedging efficacy of the contract, and the higher the likelihood of market manipulation.

To prevent a short squeeze and keep the entire menu viable, the CME assigns a static Conversion Factor (CF) to every eligible bond.

The CF acts as a price multiplier. It answers a theoretical question: "What would the price of this specific bond be if its yield-to-maturity was exactly 6%?"

**The Invoice Price**

The ultimate goal of the Conversion Factor is to calculate the Invoice Price—the actual dollar amount the long (the buyer) must pay the short (the seller) upon delivery.

The exchange strictly defines this formula:

$$Invoice\ Price = (Futures\ Settlement\ Price \times Conversion\ Factor) + Accrued\ Interest$$
>
>⚠️**Take Note**
>The conversion factor only scales the clean price.  Accrued interest is added on at the end, without being scaled by the conversion factor..

### 🔦 Peeking Inside the Black Box
The actual CME formula looks like a legacy math puzzle, but the intuition is just a simplified Present Value (PV) calculation. Rather than calculating a perfect, down-to-the-day discounted price, the exchange takes two pragmatic shortcuts to standardize the math:

1. **Rounding the Calendar:** The CME does not calculate maturity down to the exact day. Instead, it rounds the remaining time-to-maturity down to complete three-month quarters (or complete months, depending on the specific contract).

2. **The 6% Annuity:** With the dates rounded, the CME simply treats the remaining coupon payments as a standard annuity. It discounts those fixed cash flows, plus the final principal, back to the present using a strict 6% yield.

Because of these simplifications, the Conversion Factor is not a perfect reflection of reality—it is an approximation. But it accomplishes its goal. By pegging everything to a 6% baseline, the CF standardizes the basket.

**But here is the best news for a trader: you never actually have to calculate it.**

The CME runs this math and publishes the official Conversion Factors for every eligible bond well before the contract even starts trading. There is no room for argument, no bespoke modeling, and no disputing the rounding rules. The exchange's published number is absolute law.
<details>

<summary><b>🧮 Walking Through the Math: A Conversion Factor Example</b></summary>

Assume we are holding a short position in the **December 2026 Classic Bond (ZBZ6)**. We are looking at a specific deliverable bond in the secondary market: the **4.00% Treasury Bond maturing November 15, 2046**.

The official formula is a wall of algebra:

$$factor = a * \left[\left(\frac{coupon}{2}\right) + c + d\right] - b$$

If you try to read the CME rulebook, calculating the Conversion Factor for this bond feels like jumping through a maze of fractional dates and rounding tables. But if we map it directly to our three shortcuts, the logic is perfectly linear:

 **1. Rounding the Calendar**
   * Reality: From December 1, 2026, to November 15, 2046, is exactly 19 years, 11 months, and 14 days.
   * The Shortcut: The CME ignores the exact days and rounds the remaining time down to the nearest complete 3-month quarter. 19 years and 11 months becomes exactly 19 years and 9 months (79 quarters).

 **2. The 6% Annuity Engine ($c$ and $d$):** The core bracket of the equation calculates the present value of the bond. It treats the remaining principal and coupon payments as a standardized annuity, discounting them back at the strict 6.00% CME standard yield.

 **3. The Fractional Adjustment ($a$ and $b$):** Because the calendar was forcefully rounded to blunt quarters, the standard annuity formula in the middle slightly misaligns the exact coupon periods. The variables on the outside of the brackets ($a$ and $b$) simply act as fractional adjustments, mathematically smoothing those rounding errors to arrive at a clean price multiplier.
When you run this exact bond through the formula for December 2026, it spits out a final multiplier of 0.7712.

>💡 The Takeaway: Function Over Perfection
>
>Is this conversion formula a mathematically flawless Present Value calculation? No. The rounding rules create deliberate pricing inaccuracies.
>
>But it doesn't need to be perfect; it just needs to work. By successfully pegging the entire basket to a standardized 6% baseline, the Conversion Factor expands the delivery menu, prevents a short squeeze, and keeps the futures market highly liquid.
>
>And as we noted above, traders make their money precisely by hunting down the mathematical imperfections this formula leaves behind.

</details>

## 🕹️ Controlling Delivery: The CTD and Other Embedded Options
Because the short seller controls the delivery process, Treasury futures grant them three distinct advantages or "embedded options." The short seller ultimately decides what to deliver, when to deliver it, and exactly how to time the market close.

1. **The Quality Option (CTD):** The choice of which specific eligible bond from the basket to deliver.

2. **The Timing Option:** The choice of when to deliver on any business day during the delivery month. This allows the seller to hold the position if short-term financing (repo) rates are favorable, or deliver early if carrying costs become a drag.

3. **The Wildcard Option:** A structural timing quirk. The daily futures settlement price—which locks in the invoice price—is determined at 2:00 PM (Central). However, the short seller has until the exchange's 6:00 PM (Central) deadline to submit a Notice of Intention to deliver. If cash bond prices drop during that four-hour window, the short seller can snap up cheaper cash bonds and announce delivery by 6:00 PM, cashing in on the higher, locked-in 2:00 PM futures price.

All three options reduce the theoretical price of the futures contract, but their economic significance varies wildly. We will set the intraday Wildcard Option aside, as it is a fleeting execution quirk.

The Quality Option (CTD), however, dictates the vast majority of the contract's pricing, and it is inextricably linked to the Timing Option. The cheapest bond to deliver on December 1st might not be the cheapest on December 31st if interim coupons or changing carry costs shift the math mid-month.

To find the true CTD, our model will point directly at this relationship. We will calculate the total cost (Net Basis) of an Immediate Delivery (First Day) versus a Delayed Delivery (Last Day) for every eligible bond. By forcing these two timing scenarios into a head-to-head showdown, the math will automatically reveal which bond is cheapest to deliver, and exactly when to deliver it.

## 🧮 Finding the CTD: The Net Basis Calculation
In previous chapters, we used the cost-of-carry equation to calculate the forward price ($F_t$) of a stock index. You will recall the formula:
$$F_t = S_t\times\left(1+(r_t+\Delta_t)\times\frac{T-t}{360}\right) - FVD_t$$
The logic for a physical delivery Treasury future is identical. But instead of an index of stocks paying dividends, we are looking at a cash bond paying coupons. To determine which bond is the Cheapest-to-Deliver (CTD), a short seller evaluates a Cash-and-Carry trade: buy the cash bond today, hold it until the delivery date, and hand it to the exchange to settle the short position. To figure out which bond is the most efficient to deliver, we take our equity forward formula and translate it into fixed-income terms:

1. **The Forward Cash Price:** We simply swap the equity variables for bond variables to find the true, break-even forward price of holding the cash bond:
   * $S_t$: The current Spot price of the cash bond.
   * $r$: The short-term financing cost (repo rate) used to borrow money to buy the bond over the holding period ($\frac{T-t}{360}$).
   * $\Delta_t$ is the dynamic arbitrage band.
   * $FVD_t \rightarrow FVC_t$: The Future Value of Coupons. This replaces dividends, accounting for any coupon payments received while holding the bond, plus the interest earned from reinvesting them.

>The forward price on the delivery day $T$ is:

$$Forward\ Cash\ Price = S_t \times \left(1 + (r_t+\Delta_t) \times \frac{T-t}{360}\right) - FVC_t$$

2. **The Net Basis:** is the difference between what it actually costs the short seller to source and hold the bond (the Forward Cash Price) versus the revenue the exchange guarantees they will receive for it (the Invoice Price).

$$Net\ Basis = Forward\ Cash\ Price - Invoice\ Price$$

**The Goal Of the Short**

It's simple: minimize the net cost of delivery or the $Net\ Basis$. The short's pursuit of profit tethers the cash price of the **CTD** bond to the futures price. But there is slack here. As we just discussed, the short seller is granted embedded options (Quality, Timing, Wildcard). They have real economic value that isn't ignored by the market. As you likely expect, those options get competed away driving down the futures price.

### 🔗Merging Market Prices with the Eligible Bond Universe

BBefore we can calculate the cost of carry and find the cheapest-to-deliver (CTD) bond, we need to match our list of CME-eligible bonds with current market prices from FedInvest.

Financial datasets often have slight formatting quirks—such as hidden timestamps in date columns or floating-point rounding differences in coupons—that can cause matching errors. To account for these real-world data mismatches, our process follows three steps:
1. **Standardizes** the maturities and coupon rates across both datasets to ensure identical formatting.
2. **Runs a diagnostic** to ensure no eligible bonds are accidentally dropped or missing from our pricing feed.
3. **Merges** the clean data into a single, combined dataset for our basis calculations so we can begin our basis calculations.

````python
# Elgible are all bonds so drop SECURITY TYPE
clean_df = clean_df.drop(columns='SECURITY TYPE')

# 1. Standardize names and flatten
clean_df.index.name = 'Maturity'
clean_df_flat = clean_df.reset_index()

df_eligible = df_eligible.rename(columns={'Coupon Rate': 'Coupon'})
df_eligible_flat = df_eligible.reset_index()

# 2. Standardize 'Maturity' to pure dates
clean_df_flat['Maturity'] = pd.to_datetime(clean_df_flat['Maturity']).dt.normalize()
df_eligible_flat['Maturity'] = pd.to_datetime(df_eligible_flat['Maturity']).dt.normalize()

# 3. Standardize 'Coupon' to floats and round
clean_df_flat['Coupon'] = clean_df_flat['Coupon'].astype(float).round(4)
df_eligible_flat['Coupon'] = df_eligible_flat['Coupon'].astype(float).round(4)

# ---------------------------------------------------------
# Diagnostic: Find eligible bonds NOT in FedInvest
# ---------------------------------------------------------
df_check = pd.merge(
    df_eligible_flat,
    clean_df_flat[['Maturity', 'Coupon']].drop_duplicates(),
    on=['Maturity', 'Coupon'],
    how='left',
    indicator=True
)

missing_from_fedinvest = df_check[df_check['_merge'] == 'left_only'].drop(columns=['_merge'])

if not missing_from_fedinvest.empty:
    print("⚠️ WARNING: Eligible bonds missing from FedInvest:")
    display(missing_from_fedinvest)
else:
    print("✅ All eligible bonds are in FedInvest.")
# ---------------------------------------------------------

# 4. Final Inner Merge
df_combined = pd.merge(
    clean_df_flat,
    df_eligible_flat,
    on=['Maturity', 'Coupon'],
    how='inner'
)

# 5. Set index back and sort
df_combined = df_combined.set_index('Maturity').sort_values(by=['Maturity', 'Coupon'])
````

### 🩹 Bridging the Liquidity Gap: Calculating the Clean Price

Because older off-the-run bonds suffer from spotty quoting, we cannot rely strictly on the `END OF DAY` price column.

Instead, we use a vectorized approach (via `np.where`) to calculate a robust proxy for the clean price:
- **The Mid-Price (Primary):** If a valid bid and ask both exist (i.e., $> 0$), we use the mid-market price (the average of the bid and ask).
- **The Fallback:** If quotes are incomplete, we take the maximum available price across the `END OF DAY`, `Price Bid`, and `Price Ask` columns.

````python
# copy used to make sure the DataFrames aren't linked
num_rows = clean_df.shape[0]
df_ctd = df_combined[:num_rows].copy()

# vectorized calculation of clean prices
bid_ask_true = (df_ctd['Price Bid'] > 0) & (df_ctd['Price Ask'] > 0)

# bid/ask average assigned to true_value
true_value = (df_ctd['Price Bid'] + df_ctd['Price Ask']) / 2

# maximum of three prices assigned to false value
false_value = df_ctd[['END OF DAY', 'Price Bid', 'Price Ask']].max(axis=1)

# equivalent to a vectorized if-else assignment of clean price
df_ctd['Clean Price'] = np.where(bid_ask_true, true_value, false_value)
````

### 📦 Evaluating the Delivery Option

Instead of calculating the net basis for every single day in the delivery month, we can determine the optimal delivery date by comparing two daily cash flows:

1. **Daily Income:** The daily accrued interest earned by holding the bond (Daily Accrual).
2. **Daily Cost:** The daily financing cost to hold the position (Spot Dirty Price $\times$ Daily SOFR).

**The Decision Rule:**
- If **Daily Income > Daily Cost** (Positive Carry): The bond pays you more than it costs to hold. Keep it as long as possible. 👉 **Deliver on the LAST day.**
- If **Daily Income < Daily Cost** (Negative Carry): The bond costs more to hold than it yields. Stop the bleeding immediately. 👉 **Deliver on the FIRST day.**

Below, we apply this daily carry logic to our eligible bonds to determine their theoretical optimal delivery dates.

But if a bond pays a coupon *before* the last day of the delivery month, its accrued interest instantly drops to zero, and the short position receives a large cash payout that lowers their financing balance. This fractures the static calculation.

To find the true optimal delivery date, a trader must explicitly evaluate the carry economics for the periods both before and after the coupon payment. For now, we will ignore this complication. We will solve it later when we calculate the final Net Basis for each bond.

## 💵 From Clean to Dirty Bond Prices and the Delivery Decision

To calculate the invoice price of a bond, we first need to determine the accrued interest on the settlement date. The `scheduled_pay_dates` function from the `financial_quant` library returns all upcoming payment dates for a bond based on its maturity and payment frequency.

Since all eligible bonds will make a payment within the next six months, we can identify the next payment date by temporarily shifting the maturity year to the current year plus one. Once we have the next scheduled payment date, the previous payment date is simply six months prior (adjusting for month-end dates as needed).

The `timing_option` function uses these dates to calculate five key metrics for each bond:
- **Accrued Interest**
- **Dirty Price** (Clean Price + Accrued Interest)
- **Daily Accrual** (Daily Income)
- **Daily Cost of Carry** (Daily SOFR Cost)
- **Net Payment Date** (Allows us to account for an interim coupon)

````python
def timing_option(settlement, maturity, coupon, clean_price, sofr_rate, freq=2):
    """
    Calculates accrued interest, dirty price, daily carry, and the optimal delivery
    timing (Immediate vs Delay) for a Treasury bond.
    """
    settlement = pd.Timestamp(settlement)
    maturity = pd.Timestamp(maturity)
    current_year = settlement.year

    # --- HELPER FUNCTION ---
    def get_proxy_maturity(mat, target_year):
        """Creates proxy maturity, utilizing Pandas' native month-end calendar logic."""
        if mat.is_month_end:
            # Safely grab the exact last day of the target month (handles leap years natively)
            last_day = pd.Timestamp(year=target_year, month=mat.month, day=1).days_in_month
            return pd.Timestamp(year=target_year, month=mat.month, day=last_day)
        else:
            # Non-EOM dates (like the 15th) are immune to the Feb 29th ValueError
            return mat.replace(year=target_year)
    # -----------------------

    # Optimize by creating a proxy maturity date just 1 year out
    proxy_maturity = get_proxy_maturity(maturity, current_year + 1)

    # 1. Find coupon dates (fq prefix for financial_quant)
    dates = fq.scheduled_pay_dates(proxy_maturity, settlement=settlement, freq=freq)
    next_date = pd.Timestamp(dates[0])

    # Dynamically determine months between payments
    months_between = int(12 / freq)

    # 2. Handle end-of-month Treasury rule for the previous coupon
    if next_date.is_month_end:
        last_date = next_date - pd.offsets.MonthEnd(months_between)
    else:
        last_date = next_date - pd.DateOffset(months=months_between)

    # 3. Calculate Actual/Actual day counts
    days_between = (next_date - last_date).days
    days_since = (settlement - last_date).days

    # 4. Financial Calculations
    coupon_payment = coupon / freq
    accrued_interest = coupon_payment * (days_since / days_between)
    daily_accrual = coupon_payment / days_between

    dirty_price = clean_price + accrued_interest
    daily_sofr_cost = dirty_price * (sofr_rate / 360)

    return accrued_interest, dirty_price, daily_accrual, daily_sofr_cost, next_date
````

    print(f"   Daily Income: ${daily_inc:.4f} | Daily Cost:${daily_cost:.4f}")
    print(f"   Action: {decision} Delivery\n")

### 💡 A Note on the Financing Rate (SOFR)

In reality, a trader buying a bond for future delivery finances it via the repo market. To calculate the exact cost of carry, a desk would build a forward curve or use a term repo rate perfectly matched to the expected delivery date.

For this exercise, we are using a **constant, flat SOFR rate** for the entire carry period. We make this simplification for two reasons:

1. **Isolating the Core Concept:** Constructing forward term-structure curves is a complex topic of its own. Using a flat rate perfectly demonstrates the *mechanics* of the delivery decision (Income vs. Cost) without drowning the code in curve-building algorithms.
2. **The Uncertainty of the Timing Option:** Because the exact delivery date isn't locked in until the trader actually pulls the trigger, the actual term of the financing is a moving target. A constant spot rate provides a clean, stable baseline to see exactly how the timing option behaves under different conditions.

````python
# Use FredReader to the SOFR Rate
sofr_11_09=fred_data.get_series('SOFR',start_date=settlement,end_date=settlement)

# Get the single value and convert it into floating decimal rate
sofr=sofr_11_09.values[0]
sofr_rate=float(sofr[0])
sofr_rate=sofr_rate/100
````

### 🧠 The Mechanics of the Timing Flip

If a bond pays an interim coupon (e.g., on November 15th), the economics of the trade split into two distinct phases.

This creates a fascinating dynamic for our delivery timing decision:
1. **The Decision is Driven by Phase Two:** The carry cost from our Nov 9 settlement to the Nov 15 coupon date is locked in no matter what. Therefore, the decision to deliver early (Dec 1) or late (Dec 31) depends *entirely* on the new cost of carry and accrual rates established from Nov 15 onward.
2. **Dynamic Coupon Reinvestment:** Once that interim coupon is paid, we are holding cash. Every extra day we delay delivery is another day we earn SOFR interest on that cash coupon. This means our $FVC$ (Future Value of Coupons) is actually higher on Dec 31 than it is on Dec 1!

Rather than writing complex, conditional loop logic to model these two phases for every single bond, we can let Pandas do the heavy lifting. By calculating the Net Basis for both delivery dates, the math naturally accounts for the reinvestment interest and the post-coupon carry, automatically identifying the optimal timing.

## 🧾 Calculating the Invoice Price

The Invoice Price is the actual cash amount the short position receives upon delivery. It is simply the Futures Settlement Price scaled by the bond's Conversion Factor, plus the accrued interest owed on the exact day of delivery.

Because we must evaluate both a First Day (Dec 1) and Last Day (Dec 31) delivery, we will calculate two separate invoice prices.

If a bond pays a coupon before our delivery date, the accrued interest drops to zero on the payment date, and we only accrue interest for the days following the payment.*

$$Invoice\ Price = (Futures\ Settlement\ Price \times Conversion\ Factor) + Accrued\ Interest$$

We now have all the ingredients except for the $Futures\ Settlement\ Price$.

**Crucially, we need the futures settlement price on the *transaction day* (November 9, 2026), not the *settlement day* (November 10, 2026).**

To grab this price, we will use the `MASSIVEReader` instance (`massive_data`) that we created earlier in the notebook. Because we are only querying data for a single day, we will simply assign our target transaction date to both the `start_date` and `end_date` parameters of the `get_futures_data` method.

````python
# Use MASSIVEReader
# Query the December 2026 30-Year T-Bond futures (ZBZ6) on the transaction date
start='2026-11-09'
futures_11_09=massive_data.get_futures_data('ZBZ6',start_date=start,
                                           end_date=start)

# Extract the scalar settlement price from the single-row DataFrame
settlement_price=futures_11_09['settlement_price'].iloc[0]
settlement_price
````

## 🧹 Cleaning Up and Calculating the Invoice Price

Before we calculate the final Net Basis, let's tidy up our DataFrame and calculate the Invoice Price for each bond.

* **Drop Unnecessary Columns:** The raw bid, ask, and end-of-day price columns have served their purpose. We will drop them to keep our DataFrame clean and focused on the delivery economics.
* **Add the Invoice Price:** We will calculate the `Invoice_Price` for every bond simultaneously using fast, vectorized Pandas operations, applying our formula: $(Futures\ Settlement\ Price \times Conversion\ Factor) + Accrued\ Interest$.

### 📅 Setting the Delivery Window

For our December 2026 (`ZBZ6`) contract, the first and last delivery days are straightforward: **December 1** and **December 31, 2026**, both are standard business days.

**A Quick Note on Production Calendars:** Using standard weekday logic (`BDay`) is a convenient shortcut for this exercise. A production version checks against actual bank (Fedwire/SIFMA) and exchange (CME) holiday calendars—specifically accounting for **Good Friday** (a market holiday for bond settlement and futures clearing but not a federal bank holiday).

````python
# First and last business days of the delivery month (Dec 2026)
# Note: Production systems adjust for SIFMA/Fedwire market holidays
first_delivery = pd.Timestamp('2026-12-01') + pd.offsets.BDay(0)
last_delivery = pd.Timestamp('2026-12-31') - pd.offsets.BDay(0)
first_delivery, last_delivery
````

### 🗓️ A Crucial Nuance: Spot Accrued vs. Delivery Accrued

Before we calculate the Invoice Price, we must adjust our accrued interest.

The `Accrued_Interest` column we calculated earlier represents the interest owed on the **spot settlement date** (November 10). However, the Invoice Price requires the accrued interest on the **delivery date** (in December).

To bridge this gap, we will calculate the `Days_to_Delivery` and multiply it by our `Daily_Accrual`. Adding this to our original accrued interest gives us our true `Delivery_Accrued` for the Invoice Price!

````python
# 1. Define Dates
t_settlement = pd.Timestamp(settlement)

days_first = (first_delivery - t_settlement).days
days_last = (last_delivery - t_settlement).days

# 2. Check for Interim Coupons and calculate Days Since Coupon
has_interim_first = df_ctd['Next_Coupon_Date'] <= first_delivery
has_interim_last = df_ctd['Next_Coupon_Date'] <= last_delivery

days_since_coupon_first = (first_delivery - df_ctd['Next_Coupon_Date']).dt.days
days_since_coupon_last = (last_delivery - df_ctd['Next_Coupon_Date']).dt.days

# 3. Calculate Delivery Accrued (Resetting to 0 if an interim coupon was paid)
accrued_first = np.where(
    has_interim_first,
    df_ctd['Daily_Accrual'] * days_since_coupon_first,
    df_ctd['Accrued_Interest'] + (df_ctd['Daily_Accrual'] * days_first)
)

accrued_last = np.where(
    has_interim_last,
    df_ctd['Daily_Accrual'] * days_since_coupon_last,
    df_ctd['Accrued_Interest'] + (df_ctd['Daily_Accrual'] * days_last)
)

# 4. Calculate Invoice Prices
invoice_base = settlement_price * df_ctd['Conversion_Factor']
df_ctd['Invoice_First'] = invoice_base + accrued_first
df_ctd['Invoice_Last'] = invoice_base + accrued_last

# Display the Invoice Price results

#  Drop raw price columns to declutter our working DataFrame
df_clean_ctd = df_ctd.drop(columns=['Price Bid', 'Price Ask', 'END OF DAY']).copy()
display(df_clean_ctd[['Clean Price', 'Next_Coupon_Date', 'Invoice_First', 'Invoice_Last']].head())
````

## 📈 Calculating the Forward Cash Prices

Now that we know what we will be paid at delivery (Invoice Price), we need to calculate our total cost to acquire and hold the bond. This is the **Forward Cash Price** (specifically, the Forward Dirty Price).

The Forward Cash Price is simply our initial out-of-pocket cost to buy the bond today (the Spot Dirty Price) plus our total financing cost (SOFR) over the carry period. Just like the Invoice Price, we will calculate this for both our Immediate (Dec 1) and Delayed (Dec 31) delivery dates.

````python
# 1. Calculate Spot Dirty Price (Clean Price + Spot Accrued Interest)
df_clean_ctd['Spot_Dirty'] = df_clean_ctd['Clean Price'] + df_clean_ctd['Accrued_Interest']

# 2. Define the cash coupon payment (Semi-annual)
coupon_pmt = df_clean_ctd['Coupon'] / 2

# 3. Calculate Forward Cash Prices
# If an interim coupon is paid, we subtract the cash received from our total forward cost
df_clean_ctd['Forward_First'] = np.where(
    has_interim_first,
    df_clean_ctd['Spot_Dirty'] + (df_clean_ctd['Daily_Cost'] * days_first) - coupon_pmt,
    df_clean_ctd['Spot_Dirty'] + (df_clean_ctd['Daily_Cost'] * days_first)
)

df_clean_ctd['Forward_Last'] = np.where(
    has_interim_last,
    df_clean_ctd['Spot_Dirty'] + (df_clean_ctd['Daily_Cost'] * days_last) - coupon_pmt,
    df_clean_ctd['Spot_Dirty'] + (df_clean_ctd['Daily_Cost'] * days_last)
)

# Display the Forward Prices
display(df_clean_ctd[['Clean Price', 'Spot_Dirty','Daily_Cost', 'Forward_First', 'Forward_Last']].head())
````

## 🎯 Finding the CTD: The Net Basis Showdown

The final step is to calculate our **Net Basis**, which is our true economic profit or loss from the carry trade.

$$Net\ Basis = Forward\ Cash Price - Invoice\ Price$$

We will calculate the $Net\ Basis$ for both delivery dates. Rather than writing complex logic to figure out if an interim coupon changed our optimal delivery timing, we simply let Pandas take the `np.minimum()` of the two dates.

The math will naturally select the best delivery day, and the bond with the absolute lowest overall $Net\ Basis$ is our **Cheapest-to-Deliver (CTD)**!

````python
# 1. Calculate Net Basis for both dates (Forward Price - Invoice Price)
df_clean_ctd['Net_Basis_First'] = df_clean_ctd['Forward_First'] - df_clean_ctd['Invoice_First']
df_clean_ctd['Net_Basis_Last'] = df_clean_ctd['Forward_Last'] - df_clean_ctd['Invoice_Last']

# 2. THE GRAND FINALE: Find the minimum Net Basis
df_clean_ctd['Net_Basis'] = np.minimum(df_clean_ctd['Net_Basis_First'], df_clean_ctd['Net_Basis_Last'])

# 3. Tag the optimal delivery decision so we can see WHICH date won
df_clean_ctd['Optimal_Delivery'] = np.where(
    df_clean_ctd['Net_Basis_First'] <= df_clean_ctd['Net_Basis_Last'],
    'Immediate (Dec 1)',
    'Delay (Dec 31)'
)

# 4. Sort lowest to highest to reveal the true CTD bond!
df_clean_ctd = df_clean_ctd.sort_values(by='Net_Basis')


# Display the final CTD and show exactly how the minimum was chosen
display(df_clean_ctd[['Clean Price','Spot_Dirty', 'Optimal_Delivery', 'Net_Basis_First', 'Net_Basis_Last', 'Net_Basis']].head())
````

## 🔄 Closing the Loop: The Implied Repo Rate (IRR)

In our S&P 500 cash-and-carry notebook, we identified $\Delta_t$ as the **arbitrage band**—the threshold of transaction and balance sheet costs that must be overcome before a theoretical arbitrage becomes an executable trade. The CTD basis of **-0.055060** implies that we can lock in an arbitrage profit by trading the CTD against futures.

But keep in mind, our calculations assumed that $\Delta_t$ is zero. $\Delta_t$ is the difference between the actual repo rate (SOFR) and the implied repo rate (IRR). So what is the minimum $\Delta_t$ that eliminates arbitrage profits? And what is the minimum Implied Repo Rate (IRR)?

---

> **The Minimum Arbitrage Band ($\Delta^{min}_t$):**
>
> We take our Net Basis profit ($\$0.055060$), divide it by the capital required to initiate the trade (the \$94.92 Spot Dirty price), and annualize it over our 51-day holding period (Nov 9 to Dec 31) using the standard 360-day money market convention:
>
> $$\large \Delta^{min}_t = \frac{0.055060}{94.92} \times \frac{360}{51} \approx 0.0041 \text{ (41 bps)}$$

---

> **The Minimum Implied Repo Rate ($\text{IRR}^{min}$):**
>
> Because this 41 bps represents the *excess* yield generated on top of our financing costs, we simply add this minimum $\Delta^{min}_t$ back to our baseline SOFR rate to find the break-even Implied Repo Rate (IRR).
>
> $$\large \text{IRR}^{min} = 3.89\% \text{ (SOFR)} + 0.41\% \text{ (} \Delta^{min}_t \text{)} = 4.30\%$$

---

### ⚠️ A Critical Caveat: The Embedded Options
It is crucial to recognize that this 41 basis point spread represents a **minimum** threshold. Our calculations have treated the futures contract as a simple forward commitment, ignoring the embedded delivery options held by the short seller (the Quality, Timing, and Wildcard options). As previously noted, the short's options are reflected in a diminished futures price that in turn lowers the invoice amount. To prevent arbitrage, $\Delta_t$ has to compensate for the value of these options.



## 🌊 Using Bond Metrics to Measure Market Depth

Earlier, we suggested that conversion factors are key to the depth of the futures market—the mechanism that binds the futures contract to the cash market. Our list of the five bonds with the lowest Net Basis provides a perfect demonstration. If we owned any one of these bonds, would its risk profile be securely anchored by the futures contract?

We can answer this question by examining the core metrics of each bond. To that end, we will use two functions from the `financial_quant` package:

1. **`create_payoff_df`:** Creates a DataFrame mapping all the unique payment dates for the bonds. Each row represents the specific cash flows for a single bond.
2. **`calc_bond_metrics_2d`:** Calculates the Yield-to-Maturity (YTM), Modified Duration, and Convexity using that matrix of payoffs.

We will create a copy of the first five rows of `df_clean_ctd` and pass it to `create_payoff_df`. The transpose of that DataFrame, along with the bonds' cash flow maturities in years, provides the necessary inputs to calculate these metrics.

````python
# 1. Isolate our CTD candidates (make a clean copy of the top 5)
df_top5 = df_clean_ctd.head(5).copy()

# 2. Map out the future cash flows for these specific bonds
df_payoff_top5 = fq.create_payoff_df(df_top5, settlement)

# 3. Calculate the time to each cash flow in years (using Actual/365.25 convention)
mat_years = (pd.to_datetime(df_payoff_top5.columns) - pd.to_datetime(settlement)).days / 365.25

# 4. Calculate Metrics: YTM, Modified Duration, and Convexity
# We pass an initial YTM guess of 5% (0.05), transpose the cash flows, and pass the spot dirty prices
ytm, duration, convexity = fq.calc_bond_metrics_2d(
    0.05,
    df_payoff_top5.T.to_numpy(),
    mat_years,
    df_top5['Spot_Dirty'].to_numpy()
)

# 5. Assign the calculated metrics back to our top 5 DataFrame
df_top5[['YTM', 'Duration', 'Convexity']] = pd.DataFrame(
    {'YTM': ytm, 'Duration': duration, 'Convexity': convexity},
    index=df_top5.index
)

# Display the baseline risk metrics rounded to 4 decimal places
display(df_top5[['YTM', 'Duration', 'Convexity']].round(4))
````
