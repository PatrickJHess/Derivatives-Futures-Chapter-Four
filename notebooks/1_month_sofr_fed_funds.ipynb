## Two Closely Related Contracts
In the previous notebook (*Short-Term Interest Rate Contracts: The 3-Month SOFR Futures*), we documented the deep calendar liquidity of the 3-month SOFR contract as opposed to the shallow calendar liquidity of the 1-month SOFR and 30-day Fed Fund contracts. Although both contracts derive from short-term rates, their calendar liquidities point to different applications.

The 3-month contract serves the commercial term borrowing market, where loans that mature in years are linked to daily compounded values of SOFR. The 1-month SOFR and 30-day Fed Funds serve the overnight lending market. These contracts are settled using the daily average of the rates during a contract month—a measure of the overnight market that would apply to financing a Treasury security with a repo transaction.

- **1-Month SOFR (SR1):** Settles based on the arithmetic average of daily SOFR rates (including weekends and holidays) for the named contract calendar month.
- **30-Day Federal Funds (ZQ):** Settles based on the arithmetic average of the daily Effective Federal Funds Rate (EFFR) for the named contract calendar month.

## Two Highly Correlated Contracts
Our first empirical exercise demonstrates that price changes in deferred 1-month SOFR and Fed Funds futures move almost in lockstep, exhibiting a correlation of approximately 0.95. For context, this correlation is roughly the same as the correlation between the S&P 500 and Nasdaq-100 futures contracts. Yet despite this statistical convergence, the two contracts are not perfect substitutes.

The contracts reflect distinct market mechanics:
- **Fed Funds (EFFR):** An unsecured interbank lending market dominated by non-bank cash providers (primarily Federal Home Loan Banks) lending to foreign banking institutions.
- **SOFR:** A secured, multi-trillion-dollar repo market that directly finances U.S. Treasury inventory and private leveraged positions.

## From Correlation to Collateral: The Fed's Funding Labyrinth
The Fed administers three rates that together are used to maintain the SOFR rate within a narrow band:

1. **Interest on Reserve Balances (IORB):** As the name suggests, this is the rate the Fed pays for reserve deposits. By setting this rate the Fed insures that competition for arbitrage profits will cause the **Fed Funds** rate to be a few ticks below the **IOPRB**. Setting **IOPRB** provides the Fed with the necessary liquidity to intervene in the repo markets.
2. **Overnight Reverse Repo Facility (ON RRP):** The rate the Fed pays on overnight reverse repo agreements, acting as a soft floor.
3. **Standing Repo Facility (SRF):** The rate the Fed charges on overnight repo agreements, acting as a ceiling.

### What is the Fed's Objective? Facilitating Treasury Auctions
The Treasury conducts frequent, large-scale bond auctions, and the purchase of these bonds by primary dealers is primarily financed with overnight repos. Corralling the SOFR rate is meant to ensure these auctions proceed smoothly—a sudden spike in SOFR is a threat to the success of a Treasury auction.

By guaranteeing an exit floor for Treasury collateral, the central bank effectively removes a degree of market discipline. It short-circuits natural price discovery, removing the liquidity risk that would otherwise be associated with sovereign debt issuance.

The Fed's official reason for intervening in the repo market is the control of interest rates. From the perspective of a trading desk, however, the Fed works closely with the executive and legislative branches to facilitate the financing of government expenditures. This viewpoint unlocks specific trading strategies: identifying the elements that frustrate the Fed's objective.

<summary><b>💡 The EFFR Arbitrage: Why Foreign Banks Dominate "U.S. Fed Funds"</b></summary>

>*It seems counterintuitive, but the domestic Fed Funds market is driven by a regulatory loophole:*

>* **The Lenders:** Federal Home Loan Banks (FHLBs) have billions in cash but are legally barred from earning the Fed's Interest on Reserve Balances (IORB).
>* **The Domestic Penalty:** U.S. banks avoid borrowing Fed funds due to post-2008 FDIC fees assessed on total liabilities. Borrowing Fed Funds expands their balance sheet, draining any arbitrage profits earned from the **IORB/Fed funds** spread.
>* **The Foreign Loophole:** U.S. branches of Foreign Banking Organizations (FBOs) generally do not take U.S. retail deposits and pay no FDIC fees. They borrow unsecured cash from the FHLBs, park it at the Fed to earn the higher IORB rate, and pocket the risk-free spread.

## Measuring Contract Similarities

Because these contracts settle to an arithmetic average, price volatility and uncertainty naturally vanish as the front-month contract nears expiration (most of the daily rates making up the average are already locked in). To accurately capture the market's dynamic pricing of risk—rather than the mathematical certainty of the expiring average—our analysis tracks the **first deferred (second-month) contracts**.

For example, during the trading days of January 2026, we will analyze the daily price changes of the February 2026 contracts. Our observation window runs from October 1, 2025, to November 5, 2026.

We rely upon two helper functions to build this rolling dataset:
1. `get_monthly_contract_info()`: Returns the standard CME ticker suffix (e.g., 'G6' for February 2026) and calculates the exact business day boundaries for the target month.
2. `fetch_rolling_deferred_to_present()`: Iterates month-by-month, querying the deferred contract data for each trading window and efficiently concatenating the results into a single dataset.
---


> 💡 **The `clear_output()` function**
>
>`clear_output()` is a function from the IPython.display module used to remove the current output of a Jupyter Notebook cell or output widget during runtime.
> `clear_output(wait=False)`(the default) clears the output immediately.  Setting `wait=True` delays clearing until new output is ready to be displayed.


```python
def get_monthly_contract_info(date, deferred_months=0):
    """
    Returns the CME 1-month contract suffix and business days.

    Parameters:
    - date: The base observation date.
    - deferred_months: 0 = front month, 1 = second month, 2 = third month, etc.
    """
    import calendar
    anchor_date = pd.Timestamp(date)

    # 1. SHIFT THE TARGET DATE FORWARD
    # This automatically handles rolling over into new years
    target_date = anchor_date + pd.DateOffset(months=deferred_months)

    contract_year = target_date.year
    contract_month_idx = target_date.month - 1

    # 2. Map to standard CME 12-month letter codes
    month_codes = ['F', 'G', 'H', 'J', 'K', 'M', 'N', 'Q', 'U', 'V', 'X', 'Z']
    contract_code = month_codes[contract_month_idx] + str(contract_year)[-1]

    # 3. GENERATE EXACT FIRST AND LAST BUSINESS DAYS FOR THE CALENDAR MONTH
    last_day_of_month = calendar.monthrange(contract_year, target_date.month)[1]

    month_start = pd.Timestamp(year=contract_year, month=target_date.month, day=1)
    month_end = pd.Timestamp(year=contract_year, month=target_date.month, day=last_day_of_month)

    biz_days = pd.bdate_range(start=month_start, end=month_end)

    return {
        'suffix': contract_code,
        'start_date': biz_days[0],
        'end_date': biz_days[-1],
        'target_month': target_date.strftime('%Y-%m') # Added for easier debugging
    }
```

```python
def fetch_rolling_deferred_to_present(symbol,
                                      start_date,
                                      end_date=pd.Timestamp.today()):
    """
    Iterates month-by-month from the start date to the present,
    fetching the 2nd-month (deferred) contract for each trading window.
    Appends to a list and concatenates at the end for maximum performance.
    """

    start_month = pd.Timestamp(start_date).replace(day=1)
    end_date = pd.Timestamp(end_date)
    months_to_fetch = pd.date_range(start=start_month, end=end_date, freq='MS')

    # 1. Initialize the list to hold dataframes
    df_list = []

    for trade_month_start in months_to_fetch:
        # Define the trading window (stop at today if in the current month)
        trade_month_end = trade_month_start + pd.offsets.BMonthEnd(1)
        if trade_month_end > end_date:
            trade_month_end = end_date

        # Get the suffix for the contract 1 month ahead
        deferred_info = get_monthly_contract_info(trade_month_start, deferred_months=1)
        full_symbol = symbol + deferred_info['suffix']

        try:
            # Query the deferred contract over the current month's dates
            df = massive_data.get_futures_data(
                full_symbol,
                start_date=trade_month_start,
                end_date=trade_month_end
            )

            if not df.empty:
                df['Trade_Month'] = trade_month_start.strftime('%Y-%m')
                df['Contract_Target'] = deferred_info['target_month']
                df['Delta']=df['settlement_price'].diff()

                # 2. Append to the list
                df_list.append(df)

        except Exception as e:
            print(f"Skipped {full_symbol} due to API error: {e}")

    clear_output(wait=False)

    # 3. Concatenate once at the very end
    return pd.concat(df_list) if df_list else pd.DataFrame()
```

## Getting the Data and Making the Comparison
After fetching the data with our helper functions, we calculate changes in the daily settlement prices of the contracts (Delta). The describe() method in Pandas calculates the summary statistics of the contracts' deltas. The correlation of these deltas is then calculated using the corr() method


---


> 💡 **The `describe()` and `corr()` methods**
>
>The `describe()` method generates descriptive summary statistics of a DataFrame or Series. It provides a quick statistical overview of the dataset's distribution, central tendency, and dispersion.
>
>The `corr()` method computes the statistical correlation between DataFrame columns or Series objects, automatically aligning them by index.

```python
# 1. Define the observation window
start_date = '2025-10-01'
end_date = '2026-11-05'

# 2. Fetch the rolling deferred contracts for both markets
df_1_m_sofr = fetch_rolling_deferred_to_present('SR1', start_date, end_date)
df_fedfunds = fetch_rolling_deferred_to_present('ZQ', start_date, end_date)

# 3. Isolate the daily price changes (Deltas)
delta_sofr = df_1_m_sofr['Delta']
delta_fedfunds = df_fedfunds['Delta']

# 4. Generate summary statistics (distribution, central tendency, dispersion)
# Wrapping the describe() outputs in a DataFrame formats them cleanly in Jupyter
display(
    pd.DataFrame(
        [delta_sofr.describe(), delta_fedfunds.describe()],
        index=['1-Month SOFR', 'Fed Funds']
    )
)

# 5. Compute the statistical correlation of the daily price changes
# The corr() method automatically aligns the two Series by their date index
correlation = delta_sofr.corr(delta_fedfunds)
print(f'Correlation between SOFR and Fed Funds: {correlation:.4f}')
```

## 🔗 The Results Point to a Tight Relation

The summary statistics and correlation point to futures contracts that are adjacent in the risk space. The 1-Month SOFR contract displays slightly more variability than 30-Day Fed Funds, but daily changes in both are highly correlated.

Given these similarities, why do both contracts exist? What distinct risks are being expressed by having two separate markets? To begin answering this, we will next isolate the extremes: the ten highest and lowest differences in implied yields between the two contracts.


```python
# Calculate implied yield spread: (100 - P_sofr) - (100 - P_fed) = P_fed - P_sofr
sofr_fed_spread = df_fedfunds['settlement_price'] - df_1_m_sofr['settlement_price']

# 1. Isolate the extremes
minimum_spread = sofr_fed_spread.sort_values().head(10)
# Sort descending so the absolute highest spread appears at row 0
maximum_spread = sofr_fed_spread.sort_values(ascending=False).head(10)

# 2. Build the MultiIndex DataFrame directly using tuple keys
df_max_min_spread = pd.DataFrame({
    ('Minimum', 'Date'): minimum_spread.index.strftime('%Y-%m-%d'),
    ('Minimum', 'Spread'): minimum_spread.values,
    ('Maximum', 'Date'): maximum_spread.index.strftime('%Y-%m-%d'),
    ('Maximum', 'Spread'): maximum_spread.values
})

# Display the formatted table
display(df_max_min_spread)
```

## 🔧 Clues in the Plumbing: The Extremes are Not Random
Look closely at the dates in the table. The ten maximum spreads are tightly clustered in November 2025 (with one month-end outlier on October 31). The ten minimums maintain the pattern—perfectly contained within a single two-week window in May 2026.

This clustering is a big clue. In highly liquid, efficient markets, random noise is distributed smoothly over time. A tight calendar clustering points to a structural friction—a specific "leakage" in the Fed's plumbing.

Keep in mind that our analysis is with the first deferred contracts. The November 2025 extremes are actually pricing the December 2025 contracts, and the May 2026 minimums are pricing the June 2026 contracts. With this perspective, we can analyze what the futures market is anticipating:

1. **November 2025 Results (The Year-End Turn):** Regulatory scoring for global banks is calculated based on year-end balance sheet snapshots. This regulatory reporting incentivizes banks to aggressively pull back on short-term lending in December, causing secured rates like SOFR to spike as the market expects a decline in the ratio of funding to collateral.

2. **May 2026 Results (Corporate Tax Payments):** Mid-year tax payments allow the Treasury to pay down short-term T-bills, removing collateral from the repo market. With fewer assets requiring financing, the market expects an increase in the ratio of funding to collateral.

## 🚰 The 1-Month SOFR and 30-Day Fed Funds Spread and the Fed's Plumbing

We will illustrate the efficacy of these contracts by comparing their spread to the effectiveness of the Fed's plumbing. Keep in mind that the Fed's objective is to keep SOFR in ranges that make it easy to finance Treasury issuance.

How is this done? A crucial piece is the Interest on Reserve Balances (IORB). This rate attracts funds from U.S. branches of foreign banks. Competition keeps the Fed funds rate a few basis points below IORB. The IORB/Fed funds arbitrage provides funding that can be used to keep SOFR in check.

### The Where and Why of Fed Funds

*   **Where:** The Federal Home Loan Banks (FHLBs) supply nearly all the cash in the Federal Funds market.
*   **Why:** FHLBs hold massive liquidity buffers in accounts at the Federal Reserve that earn zero interest because they are legally ineligible to receive the IORB rate. Instead, they earn a return on these reserves by lending them out in the unsecured Fed Funds market.




## 🎛️ Spot Rates and Fed Plumbing Controls

We will use our `FredReader` utility to pull the historical spot rates and the Fed's administrative boundaries:

1. **SOFR:** Series ID `SOFR`
2. **Fed Funds Rate:** Series ID `DFF`
3. **Interest on Reserve Balances (IORB - Attracts Fed Funds):** Series ID `IORB`
4. **Overnight Reverse Repo Facility (Floor Rate):** Series ID `RRPONTSYAWARD`
5. **Standing Repo Facility (Ceiling Rate):** Series ID `SRFTSYD`

```python
# 1. Define the core spot rates and the Fed's administered boundary rates
series_ids = ['SOFR', 'DFF', 'IORB', 'RRPONTSYAWARD', 'SRFTSYD']

# 2. Fetch the data starting just before the November 2025 year-end squeeze
df_sofr_fed = fred_data.get_series(series_ids, start_date='2025-10-01')

# 3. Rename the cryptic FRED facility codes to clear, descriptive boundaries
# for easier plotting and analysis
df_sofr_fed.rename(
    columns={
        'RRPONTSYAWARD': 'Fed REPO Floor',   # Where the Fed takes in cash, bounding rates on the downside
        'SRFTSYD': 'Fed REPO Ceiling'        # Where the Fed provides cash, bounding rates on the upside
    },
    inplace=True
)

# 4. Clear the notebook output cell to keep the display clean after the API pull
clear_output(wait=False)
```

## 📊 Visualizing the Fed's Corridor and the Yield Spread

The function `stacked_charts` from `financial_quant` is used to demonstrate how the spread between 1-month SOFR and 30-day Fed Funds reacts to the Fed's boundaries.

The dual-panel chart below visualizes how the Fed's administrative plumbing dictates the boundaries of short-term interest rates:

* **The Top Panel (Spot Rates & Boundaries):** This chart overlays the daily spot rates for SOFR and the Fed Funds rate against the central bank's target corridor. You can clearly see the Fed's mechanics in action: the Standing Repo and Overnight Reverse Repo facilities serve as imperfect but mostly effective boundaries. Interest on Reserve Balances (IORB) acts as the arbitrage magnet for the unsecured Fed Funds rate.
* **The Bottom Panel (The Futures Spread):** Sharing the exact same timeline, the bottom chart isolates the implied yield spread between the 1-Month SOFR and 30-Day Fed Funds futures contracts.

Stacking these charts allows us to visually track how the market prices the difference between secured Treasury financing and unsecured interbank lending, and how effectively the Fed's tools contain that volatility.

Although it is difficult to notice the visual shift on the x-axis, keep in mind that the futures spread is calculated using the contract that expires in the following month. The spread in the lower panel is actively anticipating the Fed's control of SOFR.



```python
# Construct a DataFrame to hold the calculated spread and the associated contract names
df_spread = pd.DataFrame({
    'Spread': sofr_fed_spread,
    'Deferred Contract': df_1_m_sofr['Contract_Target']
})

# Extract the date indices to serve as the x-axes for both charts
top_x_axis = df_sofr_fed.index
bottom_x_data = df_spread.index

# Group the spot rates and Fed policy boundaries to plot in the top chart
top_data_list = [
    df_sofr_fed['SOFR'],
    df_sofr_fed['DFF'],
    df_sofr_fed['IORB'],
    df_sofr_fed['Fed REPO Floor'],
    df_sofr_fed['Fed REPO Ceiling']
]

# Isolate the futures spread to plot in the bottom chart
bottom_data_list = [df_spread['Spread']]

# Configure the labels, colors, and styling for the top chart
top_data_series = ['SOFR', 'Fed Funds', 'Interest On Reserve Balances',
                   'Fed REPO Floor', 'Fed REPO Ceiling']
top_colors = ['navy', 'darkorange', 'forestgreen', 'crimson', 'rebeccapurple']
top_settings = {
    'title': 'The Fed Boundaries and Spot Rates',
    'series_labels': top_data_series,
    'colors': top_colors,
    'markers': [''] * len(top_data_series) # Disables individual dot markers for clean lines
}

# Configure the labels, colors, and styling for the bottom chart
# (Note: Overwriting the initial bottom_data_series list with the string label for the legend)
bottom_data_series = ['Yield Spread']
bottom_colors = ['firebrick']
bottom_settings = {
    'title': 'Futures Spread: 1-Month SOFR Less 30-Day Fed Funds',
    'series_labels': bottom_data_series,
    'colors': bottom_colors,
    'markers': [''] * len(bottom_data_series)
}

# Establish the relative sizing ratio (top chart will be twice as tall as the bottom)
height_ratios = [2, 1]

# Pass the grouped data and settings to the custom stacked charting function
stacked_fig = fq.stacked_charts(
    top_x_axis,
    top_data_list,
    bottom_data_list,
    bottom_x_data=bottom_x_data,
    top_config=top_settings,
    bottom_config=bottom_settings
)
```

## 📈 Reading the Plumbing: Visual Evidence of Market Frictions

Looking closely at the generated plot, several structural realities of the short-term funding markets become apparent:

*   **The Unsecured Anchor:** In the top panel, notice how tightly the Fed Funds rate (orange) clings to the Interest on Reserve Balances (green). The regulatory arbitrage between the FHLBs and foreign banking organizations keeps this unsecured rate remarkably stable.
*   **The Secured Volatility:** In stark contrast, spot SOFR (navy) is more erratic, frequently testing and occasionally piercing the Fed's Standing Repo Facility ceiling (purple dashed line). A visualization of the Fed's struggles.
*   **The Forward-Looking Spread:** The bottom panel shows the futures spread peaking in late 2025. Because we are plotting the first-deferred contracts, the spread widening anticipates the massive year-end spot SOFR spikes visible in the top panel roughly one month in advance (keeping the contract's structural delay in mind).

The visual confirms our thesis: the Fed easily anchors the unsecured market, but calendar-driven collateral frictions force SOFR away from Fed Funds, creating predictable dislocations in the futures market. These disturbances are the exact recipe for trading profits and the raison d'être of the contracts!
